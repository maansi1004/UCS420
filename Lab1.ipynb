Assgn 6.1
 add all odd num from 1 to n

```python
n=int(input("Enter a number: "))
def isodd(num):
    if num%2==0:
        return False
    else:
        return True

def add_odd(n):
    s=0
    for i in range(1, n+1):
        if isodd(i):
            s+=i
    return s
print("The sum of all odd numbers from 1 to", n, "is:", add_odd(n))
```

```python
n = int(input("Enter a number: "))

def isprime(num):
    if num < 2:
        return False
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            return False
    return True

def addprime(n):
    s = 0
    for i in range(2, n + 1):
        if isprime(i):
            s += i
    return s

print("Sum of all primes from 1 to", n, "=", addprime(n))         
    
```
