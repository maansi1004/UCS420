```python
import numpy as np

# Sensor readings
temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])

# a) Add +2°C measurement error to every reading
corrected_temperature = temperature + 2

print("Original temperatures:", temperature)
print("Corrected temperatures:", corrected_temperature)


# b) Convert Celsius to Fahrenheit
fahrenheit = (9/5) * corrected_temperature + 32

print("Temperature in Fahrenheit:", fahrenheit)


# c) Identify readings greater than 32°C
readings_above_32 = corrected_temperature[corrected_temperature > 32]

print("Readings greater than 32°C:", readings_above_32)


# d) Count how many readings exceed 32°C
count = np.sum(corrected_temperature > 32)

print("Number of readings exceeding 32°C:", count)
```

```python
import numpy as np

steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

# a) Total steps
print("Total steps:", np.sum(steps))

# b) Mean
print("Mean steps:", np.mean(steps))

# c) Maximum and minimum
print("Maximum:", np.max(steps))
print("Minimum:", np.min(steps))

# d) Total for each day
print("Total for each day:", np.sum(steps, axis=0))

# e) Total for each user
print("Total for each user:", np.sum(steps, axis=1))

# f) Position of maximum
position = np.unravel_index(np.argmax(steps), steps.shape)

print("Position of maximum:", position)
print("Maximum steps:", steps[position])
```

```python
import numpy as np

X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])

y = np.array([40, 65, 30, 85])
```

```python
print("Shape:", X.shape)
print("Dimensions:", X.ndim)
```

```python
XT = X.T

print("X.T:")
print(XT)
```

```python
XTX = X.T @ X

print("X.T @ X:")
print(XTX)
```

```python
XTX_inv = np.linalg.inv(XTX)

print("Inverse:")
print(XTX_inv)
```

```python
XTX = X.T @ X
XTX_inv = np.linalg.inv(XTX)
XTy = X.T @ y

beta = XTX_inv @ XTy

print("Beta:", beta)
```

```python

# f) Coefficients
print("\nCoefficient for Sleep:", beta[0])
print("Coefficient for Activity:", beta[1])
print("Coefficient for Stress:", beta[2])


# g) Prediction
new_user = np.array([5, 40, 7], dtype=float)

predicted_score = new_user @ beta

print("\nNew user:", new_user)
print("Predicted assistance score:", predicted_score)
```
