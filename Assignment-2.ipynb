```python
def isprime(num):
    if num < 2:
        return False
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            return False
    return True
prime_numbers=[x for x in numbers if(isprime)]
print(prime_numbers)
print("Number of prime numbers", len(prime_numbers))


    
```

 iv. Delete "branch" using pop() and del
 Create two separate copies

```python
dict_pop=my_dict.copy()
dict_del=my_dict.copy()
```

```python
removed_val=dict_pop.pop("branch")
del dict_del["branch"]
```

```python

print("\niv. Using pop():")
print(dict_pop)
print("Value returned by pop():", removed_val)

print("\nUsing del:")
print(dict_del)
```

# Explanation:
 pop() removes the key AND returns its value,
 while del removes the key but does not return its value.
