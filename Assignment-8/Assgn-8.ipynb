```python
import re
import nltk
import matplotlib.pyplot as plt
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from collections import Counter
```

```python
# Download required NLTK datasets
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
```

```python
# Paragraph (6 sentences)
text = """
Artificial Intelligence is transforming the world through technology.
Machine learning allows computers to learn from data and make predictions.
Python is one of the most popular programming languages for AI development.
Deep learning helps computers understand images, speech, and natural language.
AI applications are widely used in healthcare, education, and transportation.
I enjoy learning artificial intelligence because it combines creativity and problem solving.
"""
```

```python
# 1. Convert to lowercase and remove punctuation
cleaned_text = re.sub(r'[^a-zA-Z\s]', '', text.lower())

print("Original Text:\n", text)
print("\nCleaned Text:\n", cleaned_text)
```

```python
# 2. Tokenize into words
words = cleaned_text.split()
nltk_words = word_tokenize(cleaned_text)

print("\nUsing split():")
print(words)

print("\nUsing word_tokenize():")
print(nltk_words)
```

```python
# 3. Compare split() and word_tokenize()
print("\nNumber of words using split():", len(words))
print("Number of words using word_tokenize():", len(nltk_words))
```

```python
# 4. Remove stopwords
stop_words = set(stopwords.words('english'))

filtered_words = [word for word in nltk_words if word not in stop_words]

print("\nWords after removing stopwords:")
print(filtered_words)
```

```python

# 5. Word frequency distribution
frequency = Counter(filtered_words)

print("\nWord Frequency Distribution:")
print(frequency)
```

```python

# Plot frequency distribution
plt.figure(figsize=(10, 5))
plt.bar(frequency.keys(), frequency.values())
plt.xlabel("Words")
plt.ylabel("Frequency")
plt.title("Word Frequency Distribution")
plt.xticks(rotation=45)
plt.show()
```

```python
import re
import nltk
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer

# Download datasets
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')
```

```python

# Same paragraph from Q1
text = """
Artificial Intelligence is transforming the world through technology.
Machine learning allows computers to learn from data and make predictions.
Python is one of the most popular programming languages for AI development.
Deep learning helps computers understand images, speech, and natural language.
AI applications are widely used in healthcare, education, and transportation.
I enjoy learning artificial intelligence because it combines creativity and problem solving.
"""

# 1. Extract words containing only alphabets using re.findall()
words = re.findall(r'\b[a-zA-Z]+\b', text.lower())

print("Extracted Words:")
print(words)
```

```python
# 2. Remove stopwords
stop_words = set(stopwords.words('english'))

filtered_words = [word for word in words if word not in stop_words]

print("\nAfter Stopword Removal:")
print(filtered_words)
```

```python
# 3. Stemming using PorterStemmer
stemmer = PorterStemmer()

stemmed_words = [stemmer.stem(word) for word in filtered_words]

print("\nStemmed Words:")
print(stemmed_words)
```

```python
# 4. Lemmatization using WordNetLemmatizer
lemmatizer = WordNetLemmatizer()

lemmatized_words = [lemmatizer.lemmatize(word) for word in filtered_words]

print("\nLemmatized Words:")
print(lemmatized_words)
```

```python
# 5. Compare outputs
print("\nComparison:")

for original, stem, lemma in zip(filtered_words, stemmed_words, lemmatized_words):
    print(original, " -> Stemming:", stem, "| Lemmatization:", lemma)
```

```python
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
import numpy as np
```

```python
# Three sample texts
documents = [
    "Artificial intelligence is changing the world of technology",
    "Machine learning is an important part of artificial intelligence",
    "Python is widely used for machine learning and data science"
]
```

```python
# 1. Bag of Words using CountVectorizer
count_vectorizer = CountVectorizer()

bow_matrix = count_vectorizer.fit_transform(documents)

print("Bag of Words Vocabulary:")
print(count_vectorizer.get_feature_names_out())

print("\nBag of Words Matrix:")
print(bow_matrix.toarray())
```

```python
# 2. TF-IDF using TfidfVectorizer
tfidf_vectorizer = TfidfVectorizer()

tfidf_matrix = tfidf_vectorizer.fit_transform(documents)

print("\nTF-IDF Vocabulary:")
print(tfidf_vectorizer.get_feature_names_out())

print("\nTF-IDF Matrix:")
print(tfidf_matrix.toarray())
```

```python
# 3. Find top 3 keywords from each document
feature_names = tfidf_vectorizer.get_feature_names_out()

print("\nTop 3 Keywords for Each Document:")

for i in range(len(documents)):

    scores = tfidf_matrix[i].toarray().flatten()

    top_indices = scores.argsort()[-3:][::-1]

    top_keywords = [feature_names[j] for j in top_indices]

    print(f"\nDocument {i+1}:")
    print(documents[i])
    print("Top Keywords:", top_keywords)
```

```python
import re
from nltk.tokenize import word_tokenize
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import nltk
```

```python

nltk.download('punkt')
nltk.download('punkt_tab')

# Two texts describing different technologies
text1 = """
Artificial Intelligence enables machines to perform tasks that normally require human intelligence.
It is widely used in healthcare, education, robotics, and automation.
AI systems learn from data and make intelligent decisions.
Machine learning and deep learning are important parts of artificial intelligence.
"""
```

```python
text2 = """
Blockchain is a decentralized technology used to store digital information securely.
It maintains records using distributed networks and cryptographic techniques.
Blockchain is commonly used in cryptocurrencies and secure transactions.
It provides transparency, security, and data integrity.
"""
```

```python
# 1. Preprocess and tokenize both texts
def preprocess(text):
    text = text.lower()
    text = re.sub(r'[^a-zA-Z\s]', '', text)
    tokens = word_tokenize(text)
    return tokens

tokens1 = preprocess(text1)
tokens2 = preprocess(text2)

print("Tokens of Text 1:")
print(tokens1)

print("\nTokens of Text 2:")
print(tokens2)
```

```python
# 2(a). Jaccard Similarity using sets
set1 = set(tokens1)
set2 = set(tokens2)

intersection = set1.intersection(set2)
union = set1.union(set2)

jaccard = len(intersection) / len(union)

print("\nCommon Words:", intersection)
print("Jaccard Similarity:", jaccard)
```

```python
# 2(b). Cosine Similarity using TF-IDF
documents = [text1, text2]

vectorizer = TfidfVectorizer()

tfidf_matrix = vectorizer.fit_transform(documents)

cosine_sim = cosine_similarity(tfidf_matrix[0:1], tfidf_matrix[1:2])

print("\nCosine Similarity:")
print(cosine_sim[0][0])
```

```python
# 3. Compare similarity metrics
print("\nComparison:")

print("Jaccard Similarity:", jaccard)
print("Cosine Similarity:", cosine_sim[0][0])

if cosine_sim[0][0] > jaccard:
    print("Cosine similarity gives a higher similarity score.")
else:
    print("Jaccard similarity gives a higher similarity score.")
```

```python
from nltk.sentiment import SentimentIntensityAnalyzer
from wordcloud import WordCloud
import nltk
import matplotlib.pyplot as plt
```

```python
# Download VADER lexicon
nltk.download('vader_lexicon')
```

```python
reviews=[
    "This product is very good",
    "This product has defect",
     "The quality is excellent and the performance is fantastic.",
    "I am very happy with this purchase. Highly recommended!",
    "The product is terrible and completely disappointing.",
    "Very poor quality. I regret buying this product.",
    "The product is okay. It is neither good nor bad."

]
```

```python
# Initialize sentiment analyzer
sia = SentimentIntensityAnalyzer()

positive_reviews = []
negative_reviews = []
neutral_reviews = []
```

```python
# 2. Classify reviews
print("Sentiment Analysis Results:\n")

for review in reviews:

    scores = sia.polarity_scores(review)

    compound = scores['compound']

    if compound >= 0.05:
        sentiment = "Positive"
        positive_reviews.append(review)

    elif compound <= -0.05:
        sentiment = "Negative"
        negative_reviews.append(review)

    else:
        sentiment = "Neutral"
        neutral_reviews.append(review)

    print("Review:", review)
    print("Scores:", scores)
    print("Sentiment:", sentiment)
    print()
```

```python
# Display total reviews in each category
print("Positive Reviews:", len(positive_reviews))
print("Negative Reviews:", len(negative_reviews))
print("Neutral Reviews:", len(neutral_reviews))
```

```python
# 3. Generate Word Cloud using positive reviews
positive_text = " ".join(positive_reviews)

wordcloud = WordCloud(
    width=800,
    height=400,
    background_color='white',
    colormap='viridis'
).generate(positive_text)

plt.figure(figsize=(12, 6))
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis('off')
plt.title("Word Cloud of Positive Reviews")
plt.show()
```

```python
import numpy as np
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense
```

```python
# 1. Training paragraph (~100 words)
text = """
Artificial intelligence is one of the most exciting fields of computer science.
It allows machines to learn from data and perform intelligent tasks.
Machine learning is a major part of artificial intelligence.
It helps computers identify patterns and make predictions.
Deep learning uses neural networks to solve complex problems.
Python is widely used for developing machine learning applications.
Data science combines statistics programming and domain knowledge.
Natural language processing helps computers understand human language.
Artificial intelligence is used in healthcare education finance and robotics.
Learning artificial intelligence requires practice patience and creativity.
Technology continues to evolve and create new opportunities for innovation.
Students can build intelligent applications by learning programming and mathematics.
"""
```

```python
# 2. Tokenization using Keras Tokenizer
tokenizer = Tokenizer()

tokenizer.fit_on_texts([text])

total_words = len(tokenizer.word_index) + 1

print("Total Unique Words:", total_words)

# Create input sequences
input_sequences = []

for line in text.split('.'):
    token_list = tokenizer.texts_to_sequences([line])[0]

    for i in range(1, len(token_list)):
        n_gram_sequence = token_list[:i+1]
        input_sequences.append(n_gram_sequence)

# Pad sequences
max_sequence_len = max(len(seq) for seq in input_sequences)

input_sequences = pad_sequences(
    input_sequences,
    maxlen=max_sequence_len,
    padding='pre'
)

# Separate features and labels
X = input_sequences[:, :-1]
y = input_sequences[:, -1]

# One-hot encode output labels
from tensorflow.keras.utils import to_categorical

y = to_categorical(y, num_classes=total_words)

print("\nInput Shape:", X.shape)
print("Output Shape:", y.shape)

```

```python
# 3. Build LSTM model
model = Sequential()

model.add(Embedding(
    input_dim=total_words,
    output_dim=50,
    input_length=max_sequence_len - 1
))

model.add(LSTM(100))

model.add(Dense(total_words, activation='softmax'))

model.compile(
    loss='categorical_crossentropy',
    optimizer='adam',
    metrics=['accuracy']
)

model.summary()
```

```python

# 4. Train the model
model.fit(
    X,
    y,
    epochs=150,
    verbose=1
)
```

```python
# 5. Generate new text
def generate_text(seed_text, next_words):

    for _ in range(next_words):

        token_list = tokenizer.texts_to_sequences([seed_text])[0]

        token_list = pad_sequences(
            [token_list],
            maxlen=max_sequence_len - 1,
            padding='pre'
        )

        predicted = model.predict(token_list, verbose=0)

        predicted_word_index = np.argmax(predicted, axis=-1)[0]

        output_word = ""

        for word, index in tokenizer.word_index.items():
            if index == predicted_word_index:
                output_word = word
                break

        if output_word == "":
            break

        seed_text += " " + output_word

    return seed_text


# Generate 3 new lines using seed words
print("\nGenerated Text:\n")

print("1.", generate_text("artificial", 10))
print("2.", generate_text("machine", 10))
print("3.", generate_text("technology", 10))
```
