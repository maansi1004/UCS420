```python
import cv2
from fer import FER

# Initialize emotion detector
detector = FER(mtcnn=False)
```

```python

# Define assistant responses
responses = {
    "happy": "You look cheerful! Keep smiling and enjoy your day.",
    "sad": "You seem to be showing a sad expression. Take a break and do something you enjoy.",
    "angry": "You seem to be showing an angry expression. Take a deep breath and relax.",
    "surprise": "You seem surprised! Something unexpected happened?",
    "fear": "You seem to be showing a worried expression. Take a moment to relax.",
    "disgust": "You seem to be showing an expression of dislike.",
    "neutral": "You seem to have a neutral expression. How is your day going?"
}

```

```python
# Start webcam
cap = cv2.VideoCapture(0)

if not cap.isOpened():
    print("Error: Cannot access webcam!")
    exit()

print("Emotion-Aware Cognitive Assistant Started")
print("Press Q to exit.")

while True:

    # Capture frame
    ret, frame = cap.read()

    if not ret:
        print("Unable to capture frame")
        break
```

```python
# Detect emotions
result = detector.detect_emotions(frame)

for face in result:

        # Get face coordinates
        x, y, w, h = face["box"]

        # Get emotion scores
        emotions = face["emotions"]

        # Find dominant expression
        emotion = max(emotions, key=emotions.get)

        confidence = emotions[emotion]

        # Draw rectangle around face
        cv2.rectangle(
            frame,
            (x, y),
            (x + w, y + h),
            (0, 255, 0),
            2
        )
```

```python
# Display detected expression
label = f"{emotion}: {confidence:.2f}"

cv2.putText(
            frame,
            label,
            (x, y - 10),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.8,
            (0, 255, 0),
            2
        )
```

```python
# Display assistant response
response = responses[emotion]

cv2.putText(
            frame,
            response[:55],
            (10, 30),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.55,
            (255, 255, 255),
            2
        )
print(
            "Detected Expression:",
            emotion,
            "| Confidence:",
            round(confidence, 2),
            "| Assistant:",
            response
        )
```

```python
 # Display webcam window
cv2.imshow(
        "Emotion-Aware Cognitive Assistant",
        frame
    )

    # Press Q to quit
if cv2.waitKey(1) & 0xFF == ord('q'):
    break
```

```python
# Release resources
cap.release()
cv2.destroyAllWindows()

print("Cognitive Assistant Stopped.")
```
