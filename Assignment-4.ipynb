```python
import pandas as pd
```

```python

fixed_entries = [
    {
        "question": "what is the annual fee",
        "answer": "The annual fee is Rs 500.",
        "keywords": "fee cost price charge",
        "category": "billing"
    },
    {
        "question": "how to reset password",
        "answer": "Go to Settings > Reset Password.",
        "keywords": "password reset login",
        "category": "account"
    },
    {
        "question": "what are your working hours",
        "answer": "We are open 9 AM to 5 PM.",
        "keywords": "hours timing open time",
        "category": "general"
    },
    {
        "question": "how can i pay the fee",
        "answer": "You can pay via UPI, card, or net banking.",
        "keywords": "pay payment upi fee",
        "category": "billing"
    }
]
```

```python

roll_no = input("Enter your college roll number: ")

# Extract last two digits
last_two = roll_no[-2:]

# Convert digits to integers
digit1 = int(last_two[0])
digit2 = int(last_two[1])
```

```python

categories = ["billing", "account", "general"]

category1 = categories[digit1 % 3]
category2 = categories[digit2 % 3]

print("Last two digits:", digit1, digit2)
print("Category 1:", category1)
print("Category 2:", category2)
```

```python
personalized_entries = [
    {
        "question": f"how do I update my registered mobile number",
        "answer": "You can update your registered mobile number from your account settings.",
        "keywords": "mobile number update phone account",
        "category": category1
    },
    {
        "question": f"how can I get help with my {category2} related issue",
        "answer": f"You can contact the college support desk for help with your {category2} related issue.",
        "keywords": f"{category2} help support issue",
        "category": category2
    }
]
```

```python
# 5. Combine all entries
# -------------------------------

all_entries = fixed_entries + personalized_entries

# -------------------------------
# 6. Create DataFrame
# -------------------------------

df = pd.DataFrame(all_entries)

# -------------------------------
# 7. Print final 6-row DataFrame
# -------------------------------

print("\nFinal DataFrame:")
print(df)
```

## Ques2 


```python

def score_query(query, df):
    query_words = set(query.lower().split())

    results = []

    for index, row in df.iterrows():

        # Convert keywords into words
        keyword_words = set(row["keywords"].lower().split())

        # Calculate matching words
        matched_words = query_words.intersection(keyword_words)

        # Score = number of matching keywords
        score = len(matched_words)

        if score > 0:
            results.append({
                "index": index,
                "question": row["question"],
                "answer": row["answer"],
                "category": row["category"],
                "score": score
            })

    # Sort according to confidence score
    results.sort(key=lambda x: x["score"], reverse=True)

    return results
```

```python

# Test Q2
query = input("\nQ2: Enter your query: ")

results = score_query(query, df)

print("\nMatching entries:")

if results:
    for result in results:
        print(result)
else:
    print("No matching entries found.")
```

```python

def improved_score_query(query, df):

    query_words = set(query.lower().split())

    results = []

    for index, row in df.iterrows():

        keyword_words = set(row["keywords"].lower().split())

        matched_words = query_words.intersection(keyword_words)

        score = len(matched_words)

        if score > 0:
            results.append({
                "index": index,
                "question": row["question"],
                "answer": row["answer"],
                "category": row["category"],
                "score": score
            })

    # Sort by score
    results.sort(key=lambda x: x["score"], reverse=True)

    return results
```

```python
# ==========================================
# Q6 Demonstration
# ==========================================

print("\nQ6: Tie demonstration")

tie_query = input("Enter a query for tie testing: ")

tie_results = improved_score_query(tie_query, df)

if len(tie_results) == 0:
    print("No matching entries found.")

else:
    highest_score = tie_results[0]["score"]

    # Find ALL entries having highest score
    top_results = [
        result for result in tie_results
        if result["score"] == highest_score
    ]

    print("\nHighest score:", highest_score)

    print("\nAll entries with highest score:")

    for result in top_results:
        print(result)
```
